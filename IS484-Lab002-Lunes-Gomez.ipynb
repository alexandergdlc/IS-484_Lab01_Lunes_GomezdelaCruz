# REPRESENTACIÓN DEL PROBLEMA Y BÚSQUEDA EN AMPLITUD (BFS)

## 1.1. Representación del Problema del Repartidor mediante Estados

- **Estado Inicial:** Repartidor en el nodo `"Almacen"`
- **Estado Objetivo:** Repartidor en el nodo `"ZonaNorte"`
- **Espacio de Estados:** $S = \{\text{Almacen}, \text{Centro}, \text{Terminal}, \text{Mercado}, \text{ZonaNorte}\}$
- **Acciones Posibles:** Moverse hacia un nodo adyacente directamente conectado en el mapa
- **Costo de cada acción:** 1 unidad (1 tramo recorrido)
- **Solución:** Secuencia ordenada de nodos desde el Almacén hasta la Zona Norte con el menor número de pasos

```python
# Representación del mapa de la ciudad como un grafo (diccionario de listas de adyacencia)
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print("=== REPRESENTACIÓN DEL MAPA EN PYTHON ===")
print(mapa)
```

## 1.2. Algoritmo BFS (Breadth-First Search)

BFS explora el grafo nivel por nivel (como ondas en el agua) utilizando una estructura tipo **Cola (FIFO: First In, First Out)**[cite: 1, 2]. Garantiza encontrar la ruta con menos pasos[cite: 2].

### Traza Paso a Paso (Cola FIFO)

| Turno | Camino extraído de la cola | Nodo actual | ¿Es la meta? | Se agrega a la cola |
| :---: | :--- | :---: | :---: | :--- |
| **1** | `['Almacen']` | Almacen | No | `Almacen-Centro`, `Almacen-Terminal` |
| **2** | `['Almacen', 'Centro']` | Centro | No | `Almacen-Centro-Almacen`, `Almacen-Centro-ZonaNorte`, `Almacen-Centro-Mercado` |
| **3** | `['Almacen', 'Terminal']` | Terminal | No | `Almacen-Terminal-Almacen`, `Almacen-Terminal-Mercado` |
| **4** | `['Almacen', 'Centro', 'Almacen']` | Almacen | No (Visitado) | *(Se descarta)* |
| **5** | `['Almacen', 'Centro', 'ZonaNorte']` | ZonaNorte | **Sí**  | **Camino encontrado** |

```python
def bfs(grafo: dict, inicio: str, objetivo: str):
    """
    Busca el camino más corto entre 'inicio' y 'objetivo' en un grafo usando BFS (Cola FIFO).
    
    Retorna:
        list: El camino encontrado como lista de nodos, o None si no existe ruta.
    """
    cola = [[inicio]]  # Estructura FIFO: almacena caminos por explorar
    visitados = set()  # Registro para evitar ciclos y re-evaluaciones

    while cola:
        camino = cola.pop(0)  # Desencolar el camino más antiguo (FIFO)
        nodo_actual = camino[-1]

        # Evaluación del estado objetivo
        if nodo_actual == objetivo:
            return camino

        # Expansión de estados
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo.get(nodo_actual, []):
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)

    return None

# Ejecución y salida 
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

# 2. Fundamentos Matemáticos para IA: Vectores y Matrices con NumPy

## 2.1. Representación de Datos
En Inteligencia Artificial y Machine Learning, los datos se representan mediante estructuras matriciales:
- **Vector:** Representa las características de una única observación (por ejemplo: `[temperatura, humedad]`).
- **Matriz:** Representa un conjunto de datos (*dataset*), donde cada fila es una observación y cada columna es una característica específica.

```python
import numpy as np

# Definición de observaciones individuales (Temperatura, Humedad)
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# Construcción de la matriz dataset uniendo los vectores
dataset = np.array([obs1, obs2, obs3, obs4])

print("=== MATRIZ DATASET ===")
print(dataset)
print("\nForma de la matriz (filas, columnas):", dataset.shape)
```

```python
# Acceso a elementos específicos de la matriz
print("Primera observación (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

## 2.2. Similitud y Métrica de Distancia

Para comparar qué tan parecidas son dos observaciones $\vec{a}$ y $\vec{b}$, se utilizan medidas matemáticas:

1. **Producto Escalar:** Mide la proyección de un vector sobre otro.
2. **Distancia Euclidiana:** Mide la distancia geométrica en línea recta entre dos puntos en un espacio multidimensional:

$$d(\vec{a},\vec{b}) = \sqrt{\sum_{i=1}^{n}(a_i - b_i)^2}$$

A menor distancia euclidiana, mayor es la similitud entre las observaciones. Este concepto es la base de algoritmos como **KNN** y **K-Means**.

```python
# Cálculo del producto escalar entre obs1 y obs2
producto = np.dot(obs1, obs2)

print("Producto escalar entre obs1 y obs2:", producto)
```

```python
# Fórmula manual de distancia euclidiana
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# Con la función integrada de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

## 3.1. Parte A: Búsqueda en Profundidad (DFS - Depth-First Search)

DFS explora el grafo adentrándose lo máximo posible a lo largo de cada rama antes de retroceder (*backtracking*). Utiliza una estructura de datos tipo **Pila (LIFO: Last In, First Out)**, donde el último elemento en ingresar es el primero en salir.

```python
def dfs(grafo: dict, inicio: str, objetivo: str):
    """
    Busca un camino entre 'inicio' y 'objetivo' en un grafo usando DFS (Pila LIFO).
    """
    pila = [[inicio]]  # Estructura LIFO: almacena caminos en una pila
    visitados = set()

    while pila:
        camino = pila.pop()  # Sacar el ÚLTIMO camino agregado (LIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo.get(nodo_actual, []):
                # COMPLETADO: agregar el nuevo camino a la pila LIFO
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)

    return None

# Prueba de ejecución de DFS
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("=== RESULTADO DE BÚSQUEDA EN PROFUNDIDAD (DFS) ===")
print("Camino encontrado (DFS):", resultado_dfs)
```

### Comparación de Resultados (BFS vs DFS)

- **Ruta encontrada por BFS:** `['Almacen', 'Centro', 'ZonaNorte']` (**2 pasos**).
- **Ruta encontrada por DFS:** `['Almacen', 'Terminal', 'Mercado', 'ZonaNorte']` (**3 pasos**).

**Análisis:** No encontraron el mismo camino. **BFS garantizó la ruta más corta (menor número de pasos)** debido a su exploración por capas con cola FIFO. Por su parte, DFS se adentró en la rama de `Terminal` debido al orden de desapilado LIFO, hallando una ruta válida pero no óptima.

## 3.2. Parte B: Encontrar la Observación más Cercana (Fundamento KNN)

Dada una nueva observación $v_{\text{nueva}} = [20.5, 48]$ (temperatura y humedad), calculamos la distancia euclidiana hacia cada una de las observaciones existentes en el `dataset` para identificar la más similar (menor distancia).

```python
# Nueva observación a comparar
nueva_observacion = np.array([20.5, 48])
distancias = []

# Recorrido y cálculo de distancia euclidiana para cada observación del dataset
for i, obs in enumerate(dataset):
    # COMPLETADO: distancia euclidiana con numpy
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observación {i}: {d:.4f}")

# Obtener el índice de la observación con menor distancia
indice_mas_cercano = np.argmin(distancias)
print("\nLa observación más parecida es la número:", indice_mas_cercano)
```

# 4. Parte C: Reflexión Final 

### 1. Algoritmos de Búsqueda (BFS vs. DFS)
- **BFS (Cola FIFO):** Explora por capas nivel por nivel y **garantiza encontrar la ruta óptima con menos pasos** en grafos de costo uniforme. Es ideal para problemas de ruteo como el del repartidor.
- **DFS (Pila LIFO):** Explora una rama en profundidad antes de hacer *backtracking*. Puede encontrar una solución usando menos memoria en ciertos espacios, pero **no garantiza el camino más corto** y suele desviarse por rutas más largas.

### 2. Fundamentos Matemáticos con NumPy y Similitud
- Representar observaciones como **vectores** y el dataset como una **matriz** permite vectorizar cálculos sin recurrir a bucles lentos en Python puro.
- El **producto escalar** cuantifica la alineación/proyección entre vectores, mientras que la **distancia euclidiana** mide la cercanía geométrica en línea recta entre observaciones.

### 3. Justificación y Limitaciones de la Distancia Euclidiana
- **¿Por qué es una medida razonable?** Porque interpreta el conjunto de características como coordenadas en un espacio multidimensional; a menor distancia física entre dos puntos, mayor es su similitud. Este principio fundamenta algoritmos como **KNN** y **K-Means**.
- **¿Cuándo falla o no es la mejor opción?** Falla cuando las características poseen **escalas numéricas muy distintas** (por ejemplo, *Soles/Ingresos* en miles frente a *Edad* en años). La variable con mayor magnitud dominará completamente la suma de diferencias al cuadrado, opacando a las demás. Para evitarlo, se debe aplicar **normalización o estandarización** de datos previa.
