## Escalamiento de tickets de soporte técnico

### Ficha PEAS
| Elemento | Descripción |
|---|---|
| **Percepción (S)** | `tiempo_espera_minutos` (int), `nivel_urgencia` ("baja", "media", "alta"), `cliente_premium` (Boolean) |
| **Acciones (A)** | "nivel 1", "escalar a nivel 2", "escalar a nivel 3" |
| **Entorno (E)** | Sistema de gestión de mesa de ayuda (Helpdesk) y tickets de soporte |
| **Objetivo** | Resolver incidentes técnicos cumpliendo los acuerdos de nivel de servicio (SLA) |
| **Medida de Desempeño (P)** | Tiempo medio de resolución, nivel de satisfacción (CSAT) y uso eficiente de especialistas |

### Justificación de las reglas
Las reglas garantizan una atención proporcionada al riesgo operativo y a la prioridad del usuario. Un ticket con urgencia "alta" proveniente de un cliente premium o con más de 15 minutos de espera escala inmediatamente a Nivel 3. Tickets de urgencia "media" o "baja" que superen los umbrales razonables de espera son derivados a Nivel 2 para evitar el incumplimiento de SLA y la insatisfacción del cliente.

```python
### Código
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    """
    Agente de asignación y escalamiento de tickets de soporte.
    Retorna tupla: (accion, motivo)
    """
    urgencia = nivel_urgencia.lower().strip()
    
    if urgencia == "alta":
        if cliente_premium or tiempo_espera_minutos >= 15:
            return ("escalar a nivel 3", "urgencia alta priorizada por cliente premium o espera mayor a 15 minutos")
        else:
            return ("escalar a nivel 2", "urgencia alta reportada requiere atención por especialista de nivel 2")
            
    elif urgencia == "media":
        if cliente_premium or tiempo_espera_minutos >= 30:
            return ("escalar a nivel 2", "urgencia media que excede límite de espera o proviene de cliente premium")
        else:
            return ("nivel 1", "urgencia media dentro de rango de atención estándar")
            
    else:  # "baja"
        if tiempo_espera_minutos >= 60:
            return ("escalar a nivel 2", "urgencia baja pero con tiempo de espera crítico (>= 60 min)")
        elif cliente_premium and tiempo_espera_minutos >= 30:
            return ("escalar a nivel 2", "cliente premium con tiempo de espera prolongado en urgencia baja")
        else:
            return ("nivel 1", "urgencia baja atendida en flujo normal de atención")
```

```python
### Simulación y pruebas
casos_prueba_soporte = [
    (10, "alta", True),    # Nivel 3 (Alta + Premium)
    (5,  "alta", False),   # Nivel 2 (Alta regular)
    (40, "media", False),  # Nivel 2 (Media + Espera > 30 min)
    (10, "media", False),  # Nivel 1 (Media normal)
    (10, "baja", True),    # Nivel 1 (Baja Premium sin demora)
    (65, "baja", False)    # Nivel 2 (Baja con tiempo crítico > 60 min)
]

print("==========================================================================")
print("             EVALUACIÓN DE CASOS DE PRUEBA DE SOPORTE TÉCNICO             ")
print("==========================================================================")
for espera, urgencia, premium in casos_prueba_soporte:
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"Espera: {espera:2d} min | Urgencia: {urgencia:5s} | Premium: {str(premium):5s} -> Decisión: {accion.upper()}")
    print(f"   Motivo: {motivo}\n")
```
