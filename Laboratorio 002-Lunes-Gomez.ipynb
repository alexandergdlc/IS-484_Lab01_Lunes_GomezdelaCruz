## 1. Representación del Problema mediante Estados
Todo problema de resolución mediante búsqueda en Inteligencia Artificial requiere una formulación matemática y lógica formal compuesta por los siguientes elementos principales:

- **Estado Inicial ($S_0$):** Configuración de partida en la que se encuentra el agente.
- **Estado Objetivo ($S_g$):** Configuración deseada que el agente debe alcanzar.
- **Espacio de Estados ($S$):** Conjunto finito de todas las situaciones o posiciones posibles dentro del entorno.
- **Acciones / Operadores ($A$):** Movimientos o transiciones válidas para pasar de un estado a otro.
- **Función de Costo ($c(s, a, s')$):** Valor numérico asignado a la ejecución de una acción (en este caso, 1 paso por cada desplazamiento).
- **Solución:** Secuencia ordenada de acciones que lleva del estado inicial al objetivo con costo mínimo o aceptable.

### Ficha Formal del Problema: Ruteo del Repartidor

| Componente | Descripción Formal |
| :--- | :--- |
| **Estado Inicial** | Repartidor posicionado en el nodo `"Almacen"` |
| **Estado Objetivo** | Repartidor posicionado en el nodo `"ZonaNorte"` |
| **Espacio de Estados** | $S = \{\text{Almacen}, \text{Centro}, \text{Terminal}, \text{Mercado}, \text{ZonaNorte}\}$ |
| **Acciones Posibles** | Transitar hacia un nodo vecino directamente conectado por una arista en el grafo de la ciudad |
| **Costo por Acción** | $c(s, a) = 1$ unidad (1 paso por tramo recorrido) |
| **Criterio de Optimidad** | Encontrar el camino que minimice el número total de transiciones (pasos) |

```python
# ==============================================================================
# REPRESENTACIÓN DEL MAPA MEDIANTE UN GRAFO (DICCIONARIO DE LISTAS DE ADYACENCIA)
# ==============================================================================

# Definición del mapa de la ciudad como un diccionario de listas
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"]
}

print("=== REPRESENTACIÓN DEL GRAFO EN PYTHON ===")
for nodo, vecinos in mapa.items():
    print(f"Nodo '{nodo}' conectado con: {vecinos}")
```

## 2. Algoritmo de Búsqueda en Amplitud (BFS - Breadth-First Search)

### Idea Central
BFS explora el espacio de búsqueda por **anillos concéntricos** o capas, de forma análoga a las ondas que se propagan en el agua al arrojar una piedra. Primero se evalúan todos los vecinos directos del punto de origen, luego los vecinos de estos, y así sucesivamente.

### Propiedades Clave de BFS:
1. **Estructura de Datos:** Utiliza una **Cola (Queue)** bajo la regla **FIFO** (*First In, First Out*: el primer camino en entrar es el primero en ser procesado).
2. **Garantía de Optimidad:** En grafos con costos uniformes (donde cada arista cuesta 1), BFS **garantiza encontrar el camino más corto** en términos de número de pasos.
3. **Manejo de Ciclos:** Mantiene un conjunto de nodos `visitados` para evitar bucles infinitos y re-exploraciones innecesarias.
